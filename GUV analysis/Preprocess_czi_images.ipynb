```python
# This code processes the dimensions of the input microscopy files (CZI format) and saves the files as TIF.
# This code was set up with the aid of ChatGPT (chatgpt.com).
import os
import numpy as np
from czifile import CziFile
import tifffile

# === Path to directory with .czi files ===
input_folder = "xxx"

# === Create output directory ===
output_folder = os.path.join(input_folder, "tiff_output")
os.makedirs(output_folder, exist_ok=True)

# === Iterate over all CZI files ===
for filename in os.listdir(input_folder):
    if filename.lower().endswith(".czi"):
        czi_path = os.path.join(input_folder, filename)
        
        print(f"Processing: {filename}")
        
        with CziFile(czi_path) as czi:
            image = czi.asarray()
        
        # === Reduce dimensions ===
        image = np.squeeze(image)
        
        # Optional debug output
        print(f"Shape after squeeze: {image.shape}")
        
        # === File name for TIFF ===
        base_name = os.path.splitext(filename)[0]
        tiff_path = os.path.join(output_folder, base_name + ".tif")
        
        # === Speichern ===
        tifffile.imwrite(tiff_path, image)
        
print("Done.")
```
